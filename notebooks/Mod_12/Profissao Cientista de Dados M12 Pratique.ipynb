Abaixo temos um dataframe com as seguintes informações:


**Nome:** O nome de cada indivíduo da base


**Idade:** A idade de cada um deles


**Profissão:** A atual ocupação


**Salário:** O salário registrado


**Limite de crédito:** O limite de crédito disponível para cada um dos indivíduos


**Historico_Inadimplencia:** Marco como 0 aqueles que nunca tiveram um histórico de inadimplência e como 1 aqueles que já tiveram em algum momento.


**Estado civil:** O estado civil de cada um deles


**Imovel_Proprio:** Marcado como 0 aqueles que não tem imóvel próprio e como 1 aqueles que tem.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python

data = {
    'Nome': ['Alice', 'Joao', 'Charlie', 'David', 'Eva', 'Diego', 'Denize', 'Claudio'],
    'Idade': [25, 30, 35, 40, 45, 60, 22, 24],
    'Profissão': ['Engenheiro', 'Médico', 'Professor', 'Advogado', 'Médico','Engenheiro', 'Estudante','Estudante'],
    'Salário': ['4500', '8000', '5000', '10000', '12000','15000', '1200','1500'],
    'Limite_Credito': ['2500', '4000', '4000', '1000', '10000','2000', '500','250'],
    'Historico_Inadimplencia': ['0', '0', '0', '1', '0','1', '0','1'],
    'Estado_Civil': ['Casamento', 'Casamento', 'Solteiro', 'Solteiro', 'Casamento','Solteiro', 'Solteiro','Solteiro'],
    'Imovel_Proprio': ['0', '0', '0', '1', '1','1', '0','0']
}

df = pd.DataFrame(data)
df.head(5)
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['Idade'], df['Salário'], color='purple', alpha=0.7)
plt.title('Relação entre Salário e Idade', fontsize=16)
plt.xlabel('Idade', fontsize=12)
plt.ylabel('Salário', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

```

```python
plt.figure(figsize=(10, 6))
# Fix the groupby operation - select only one column at a time
# If you want to group by 'Profissão' and get the mean of 'Limite_Credito'
Limite_de_Credito = df.groupby('Profissão')['Limite_Credito'].mean()
# Or if you want to use both columns in some way, you might need:
# Limite_de_Credito = df.groupby('Profissão')[['Salário', 'Limite_Credito']].mean()

plt.bar(Limite_de_Credito.index, Limite_de_Credito, color='skyblue')
for i, valor in enumerate(Limite_de_Credito):
    plt.text(i, valor, str(valor), ha='center', va='bottom', fontsize=10)
plt.title('Limite de Crédito', fontsize=16)
# Fixed duplicate xlabel - changed first one to ylabel
plt.ylabel('Índice', fontsize=12)
plt.xlabel('Salário', fontsize=12)
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
           
```

```python
# Criando o bloxplot apenas para coluna salario
plt.figure(figsize=(8, 6))
plt.boxplot(df['Salário'])

# legandas
plt.title('Boxplot do Salário', fontsize=16)
plt.ylabel('Salário', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
           
```

```python
# First, ensure the Limite_Credito column is numeric
df['Limite_Credito'] = pd.to_numeric(df['Limite_Credito'], errors='coerce')

# Now create the boxplot
plt.figure(figsize=(8, 6))
plt.boxplot(df['Limite_Credito'].dropna())  # Drop NaN values that might have been created

# legends
plt.title('Boxplot do Limite de Crédito', fontsize=16)
plt.ylabel('Limite_Credito', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
