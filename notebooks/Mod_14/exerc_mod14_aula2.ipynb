# EXEMPLO 2 - CIDADE

Outros problemas que costuma ser frequente na qualidade dos dados:
- Acentuação
- Erros de Digitação
- Letras Maiúsculas e Minúsculas

```python
dados = {
    'Idade': [ 25, 35, 45, 400, 55, 544, 60, 234, 12],
    'Cidade': [ 'São Paulo', 'Salvador', 'Rio de Janeiro', 'Curitiba', 'Salvador', 'Sao Paulo', 'salvador', 'Curitiba', 'Rio de Janeiro']
}
df = pd.DataFrame(dados)
print(df)
```

Vamos trazer a média das idades por cidade realizando um agrupamento pod cidades como vimos no módulo de etatística básica.

```python
df.groupby('Cidade')['Idade'].mean().reset_index()
```

Neste exemplo, notamos que as ciades que deveriam estar agrupadas acabam se repetindo e alguns erros de digotação e padronização dos nomes.
Se não realizarmos o tratamento desses dados dificilmente chegaremos a média correta das idades por cidade.

# EXEMPO 3 - SALÁRIO

Muitas vezes também acabamos lidando com  valores ausentes nas nosas bases de dados. Isso acontece pois muitos clientes acabam não fornecendo todos seus daos ou até mesmo por conta de falhas na coleta de dados. 

```python
dados = {
    'Idade': [25, 45, 30, 35, 400, 50, 544, 60, 65, 24, 233, 70, 36],
    'Cidade': ['São Paulo', 'Salvador', 'Rio de Janeiro', 'Curitiba', 'Salvador', 'Sao Paulo', 'salvador', 'Curitiba', 'Rio de Janeiro', 'São Paulo', 'Salvador', 'Rio de Janeiro', 'Curitiba'],
    'Salario': ['1200,0', '0', '4000,0', '500,0','0','0','0', '0','3000,0', '2500,0','0','0','1600,0'] 
}

df = pd.DataFrame(dados)
print(df)
```

Neste outro exemplo, o dataframe acima, algumas linhas estáo sem preenchimentos dos dados de sálario, retornando 0. EM uma primeira podemos até pensar quepor ter outras linhas preenciidas os valores faltantes podem não afetar sua analise.
Mas notem que algumas cidades como Salvador não tem valores de salário. Como sabemos alguns modelos de machine learning de aprendizado supervisionado utilizam daos históricos para aprender com esses dados. Imaginem que nosso modelo tivesse a proposta de prever exatamente o salário apenas analisando a idade e cidade?. Neste casso se o modelo não tiver exemplos para a cidade de Salvador isso pode enviezar os resultados sempre que essa cidade aparecer. O modelo trará uma conclusão distorcida de que sempre nessa ciade o salário será 0.
Outro fator é que quanto mais dados ausentes, masi dificultosa será a aprendizagem do modelo. O modelo precisa do máximo de informação possível.
