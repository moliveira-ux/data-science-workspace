```python
#Seu código aqui fig = px.histogram(df, x='Churn', color='Genero', barmode='stack')

import plotly.express as px

fig = px.histogram(
    df,
    x='Genero',
    color='Churn',
    barmode='stack'
)

fig.update_layout(
    title='Relação entre Churn e Genero',
    xaxis_title='Genero',
    yaxis_title='Contagem',
    legend_title='Churn'
)

fig.show()
```

R.
#Analisando gráfico acima, a relação entre as variáveis Genero e Churn, não vemos evidências visuais na taxa de Churn em relação ao genero do cliente. Pois ambos estão equilibrados.

```python
import plotly.express as px

fig = px.histogram(
    df,
    x='Idade',
    color='Churn',
    barmode='stack'
)

fig.update_layout(
    title='Relação entre Idade e Churn',
    xaxis_title='Idade',
    yaxis_title='Contagem',
    legend_title='Churn'
)

fig.show()
```

R.
#Analisando gráfico acima (modelo empilhado), a relação entre as variáveis Idade e Churn, não vemos também evidências visuais de que clientes com maior idade tendem a dar o Churn. Mas nos mostra onde estão concentrados o maior número de clientes que permanecem e qual a faixa etaria.

```python
taxa_idade = df.groupby('Idade')['Churn'].mean().reset_index()

import plotly.express as px

fig = px.line(
    taxa_idade,
    x='Idade',
    y='Churn',
    title='Taxa de Churn por Idade'
)

fig.show()
```

R.#Analisando estas mesmas variáveis, utilizando o um gráfico de linhas, podemos responder por exemplo: "À medida que a idade aumenta, o churn aumenta ou diminui?"
Neste caso, a idade possui forte influência no churn, com clientes mais velhos apresentando taxas de cancelamento maiores.
